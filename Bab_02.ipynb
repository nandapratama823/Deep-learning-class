## Machine learning dan deep learning

Machine learning mencari parameter atau aturan dari data, sedangkan deep learning memakai banyak transformasi berlapis untuk mempelajari representasi. Kata deep merujuk pada panjang jalur komputasi, bukan jaminan bahwa model lebih cerdas.


## Supervised dan unsupervised learning

Supervised learning mempelajari hubungan input-target, misalnya gambar dan label. Unsupervised learning mencari struktur tanpa target eksplisit, seperti kelompok atau representasi padat. Evaluasi keduanya berbeda: supervised dapat memakai data berlabel yang ditahan, sedangkan unsupervised memerlukan ukuran internal dan validasi domain.


## Parametrik dan nonparametrik

Model parametrik merangkum pengetahuan ke dalam parameter berukuran tetap, misalnya bobot jaringan. Metode nonparametrik dapat menyimpan atau menggunakan data pelatihan secara langsung. Istilah nonparametrik bukan berarti tanpa parameter; kompleksitas efektifnya dapat tumbuh bersama data.


## Predict, compare, learn

Paradigma utama buku adalah predict, compare, learn. Model menghasilkan prediksi, fungsi loss membandingkannya dengan target, lalu sinyal error dipakai untuk menyesuaikan bobot. Data validasi dan uji tidak boleh ikut menentukan pembaruan bobot.


## Reproduksi kode dan eksperimen

Jalankan sel dari atas ke bawah. `MODE_CEPAT=True` menjaga waktu eksekusi singkat; ubah ke `False` untuk konfigurasi yang lebih panjang.


```python
from pathlib import Path
import numpy as np

np.set_printoptions(precision=4, suppress=True)
np.random.seed(1)
DATA_DIR = (Path.cwd().parent / "data").resolve()
MODE_CEPAT = True  # ubah ke False untuk eksperimen lebih panjang
assert DATA_DIR.is_dir(), f"Folder data tidak ditemukan: {DATA_DIR}"
print("Data:", DATA_DIR)
print("Mode cepat:", MODE_CEPAT)

def sigmoid(x):
    x = np.clip(x, -60, 60)
    return 1 / (1 + np.exp(-x))

def softmax(x):
    x = np.asarray(x)
    shifted = x - np.max(x, axis=-1, keepdims=True)
    exp = np.exp(shifted)
    return exp / exp.sum(axis=-1, keepdims=True)
```

```python
# Supervised parametric learning: regresi linear dengan gradient descent
x = np.linspace(-1, 1, 40).reshape(-1, 1)
y = 2.5 * x[:, 0] - 0.3
w, b, alpha = np.zeros(1), 0.0, 0.1
for _ in range(100):
    pred = x @ w + b
    delta = pred - y
    w -= alpha * (x.T @ delta) / len(x)
    b -= alpha * delta.mean()
print("w, b, mse:", w[0], b, np.mean((x @ w + b - y) ** 2))
```

```python
# Unsupervised learning: dua centroid sederhana
points = np.array([[-1.2, -0.8], [-0.7, -1.1], [0.8, 1.0], [1.2, 0.7]])
centroids = points[[0, 2]].copy()
for _ in range(5):
    distance = ((points[:, None] - centroids[None, :]) ** 2).sum(axis=2)
    cluster = distance.argmin(axis=1)
    centroids = np.vstack([points[cluster == k].mean(axis=0) for k in range(2)])
print("cluster:", cluster)
print("centroids:", centroids)
```
