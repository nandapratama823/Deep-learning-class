## Reproduksi kode dan eksperimen

Jalankan sel dari atas ke bawah. `MODE_CEPAT=True` menjaga waktu eksekusi singkat; ubah ke `False` untuk konfigurasi yang lebih panjang.


```python
from pathlib import Path
import numpy as np

np.set_printoptions(precision=4, suppress=True)
np.random.seed(1)
DATA_DIR = (Path.cwd().parent / "data").resolve()
MODE_CEPAT = True  # ubah ke False untuk eksperimen lebih panjang
assert DATA_DIR.is_dir(), f"Folder data tidak ditemukan: {DATA_DIR}"
print("Data:", DATA_DIR)
print("Mode cepat:", MODE_CEPAT)

def sigmoid(x):
    x = np.clip(x, -60, 60)
    return 1 / (1 + np.exp(-x))

def softmax(x):
    x = np.asarray(x)
    shifted = x - np.max(x, axis=-1, keepdims=True)
    exp = np.exp(shifted)
    return exp / exp.sum(axis=-1, keepdims=True)
```

```python
architecture = [(3, 4), (4, 2), (2, 1)]
batch = 5
shape = (batch, architecture[0][0])
print("input", shape)
for index, (fan_in, fan_out) in enumerate(architecture, 1):
    assert shape[1] == fan_in
    shape = (shape[0], fan_out)
    print(f"layer {index}: W=({fan_in}, {fan_out}) -> activation={shape}")
```

```python
# Forward pass yang bentuknya sesuai diagram
rng = np.random.default_rng(7)
x = rng.normal(size=(5, 3))
weights = [rng.normal(scale=0.2, size=s) for s in architecture]
activations = [x]
for i, w in enumerate(weights):
    value = activations[-1] @ w
    activations.append(np.maximum(0, value) if i < len(weights)-1 else value)
print([a.shape for a in activations])
```
