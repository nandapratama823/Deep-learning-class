## Predict, compare, learn

Learning dimulai dengan prediksi, pengukuran error, lalu koreksi bobot. Squared error $(\hat y-y)^2$ selalu nonnegatif dan memberi penalti lebih besar pada kesalahan besar. Loss adalah tujuan optimisasi, bukan selalu metrik bisnis akhir.


## Turunan dan gradien

Turunan mengukur perubahan lokal loss terhadap bobot. Untuk neuron linear dengan squared error, gradien sebanding dengan $(\hat y-y)x$. Tanda gradien menentukan arah dan besar absolutnya menentukan sensitivitas lokal.


## Learning rate

Pembaruan $w\leftarrow w-\alpha\nabla_wL$ memakai learning rate $\alpha$. Nilai terlalu kecil memperlambat belajar; nilai terlalu besar dapat melewati minimum dan divergen. Kurva loss membantu mendeteksi masalah ini.


## Memorization dan generalization

Loss training yang kecil hanya menunjukkan kecocokan pada contoh yang telah dilihat. Tujuan sebenarnya adalah performa pada data baru. Karena itu evaluasi harus memakai data terpisah dan mempertimbangkan overfitting.


## Reproduksi kode dan eksperimen

Jalankan sel dari atas ke bawah. `MODE_CEPAT=True` menjaga waktu eksekusi singkat; ubah ke `False` untuk konfigurasi yang lebih panjang.


```python
from pathlib import Path
import numpy as np

np.set_printoptions(precision=4, suppress=True)
np.random.seed(1)
DATA_DIR = (Path.cwd().parent / "data").resolve()
MODE_CEPAT = True  # ubah ke False untuk eksperimen lebih panjang
assert DATA_DIR.is_dir(), f"Folder data tidak ditemukan: {DATA_DIR}"
print("Data:", DATA_DIR)
print("Mode cepat:", MODE_CEPAT)

def sigmoid(x):
    x = np.clip(x, -60, 60)
    return 1 / (1 + np.exp(-x))

def softmax(x):
    x = np.asarray(x)
    shifted = x - np.max(x, axis=-1, keepdims=True)
    exp = np.exp(shifted)
    return exp / exp.sum(axis=-1, keepdims=True)
```

```python
weight, goal, input_value = 0.5, 0.8, 0.5
alpha = 0.1
history = []
for step in range(20):
    prediction = input_value * weight
    error = (prediction - goal) ** 2
    gradient = 2 * (prediction - goal) * input_value
    weight -= alpha * gradient
    history.append(error)
print("weight akhir:", weight, "error awal/akhir:", history[0], history[-1])
```

```python
# Learning rate terlalu besar dapat menyebabkan divergensi
def train_scalar(alpha, steps=12):
    w, errors = 0.5, []
    for _ in range(steps):
        pred = 2.0 * w
        errors.append((pred - 0.8) ** 2)
        w -= alpha * 2 * (pred - 0.8) * 2.0
    return np.array(errors)

for rate in [0.01, 0.1, 0.6]:
    errors = train_scalar(rate)
    print(rate, "->", np.round(errors[[0, -1]], 6))
```
