## Gradien untuk banyak bobot

Jika satu prediksi menerima banyak fitur, setiap bobot memperoleh koreksi yang sebanding dengan nilai fitur dan delta output. Dalam notasi vektor, gradien adalah outer product antara input dan delta ketika terdapat banyak output.


## Credit assignment

Model harus menentukan bobot mana yang bertanggung jawab terhadap error. Fitur bernilai nol menerima gradien nol pada langkah itu; fitur besar dapat menghasilkan pembaruan besar sehingga skala fitur berpengaruh pada optimisasi.


## Banyak input dan output

Matriks bobot memetakan beberapa input ke beberapa prediksi. Bentuk gradien harus sama dengan bentuk bobot. Pemeriksaan `shape` dan pengujian numerik gradien membantu memastikan implementasi benar.


## Interpretasi bobot

Bobot menangkap korelasi yang berguna untuk menurunkan loss, bukan hubungan kausal. Bobot juga dipengaruhi skala, korelasi antarf fitur, inisialisasi, dan data yang tersedia.


## Reproduksi kode dan eksperimen

Jalankan sel dari atas ke bawah. `MODE_CEPAT=True` menjaga waktu eksekusi singkat; ubah ke `False` untuk konfigurasi yang lebih panjang.


```python
from pathlib import Path
import numpy as np

np.set_printoptions(precision=4, suppress=True)
np.random.seed(1)
DATA_DIR = (Path.cwd().parent / "data").resolve()
MODE_CEPAT = True  # ubah ke False untuk eksperimen lebih panjang
assert DATA_DIR.is_dir(), f"Folder data tidak ditemukan: {DATA_DIR}"
print("Data:", DATA_DIR)
print("Mode cepat:", MODE_CEPAT)

def sigmoid(x):
    x = np.clip(x, -60, 60)
    return 1 / (1 + np.exp(-x))

def softmax(x):
    x = np.asarray(x)
    shifted = x - np.max(x, axis=-1, keepdims=True)
    exp = np.exp(shifted)
    return exp / exp.sum(axis=-1, keepdims=True)
```

```python
inputs = np.array([8.5, 0.65, 1.2])
weights = np.array([0.1, 0.2, -0.1])
target, alpha = 1.0, 0.01
for _ in range(20):
    prediction = inputs @ weights
    delta = prediction - target
    weight_delta = inputs * delta
    weights -= alpha * weight_delta
print("prediction:", inputs @ weights, "weights:", weights)
```

```python
# Banyak input dan output: gradien berbentuk outer product
inputs = np.array([0.5, 1.0, -0.5])
targets = np.array([1.0, 0.0])
weights = np.array([[0.1, -0.2], [0.0, 0.3], [0.2, 0.1]])
prediction = inputs @ weights
delta = prediction - targets
gradient = np.outer(inputs, delta)
assert gradient.shape == weights.shape
weights -= 0.1 * gradient
print("prediksi sebelum/sesudah:", prediction, inputs @ weights)
```
