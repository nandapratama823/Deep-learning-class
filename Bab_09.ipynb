## Peran aktivasi

Aktivasi nonlinear membuat jaringan bertumpuk lebih ekspresif daripada satu transformasi linear. Pilihan aktivasi juga memengaruhi skala aktivasi dan kestabilan gradien.


## Hidden-layer activations

ReLU sederhana dan efisien tetapi dapat menghasilkan unit mati. Tanh berpusat di nol dan terbatas pada [-1,1], namun dapat jenuh. Sigmoid cocok untuk probabilitas biner tetapi mudah jenuh pada nilai ekstrem.


## Output activations

Sigmoid memodelkan peluang biner, sedangkan softmax mengubah sekumpulan logit menjadi distribusi kategori yang berjumlah satu. Implementasi softmax stabil mengurangi nilai maksimum sebelum eksponensial.


## Turunan dan loss

Backprop mengalikan delta dengan turunan aktivasi. Kombinasi softmax dan cross-entropy menghasilkan gradien logit yang ringkas, yaitu prediksi dikurangi target, sekaligus menghindari beberapa masalah numerik.


## Reproduksi kode dan eksperimen

Jalankan sel dari atas ke bawah. `MODE_CEPAT=True` menjaga waktu eksekusi singkat; ubah ke `False` untuk konfigurasi yang lebih panjang.


```python
from pathlib import Path
import numpy as np

np.set_printoptions(precision=4, suppress=True)
np.random.seed(1)
DATA_DIR = (Path.cwd().parent / "data").resolve()
MODE_CEPAT = True  # ubah ke False untuk eksperimen lebih panjang
assert DATA_DIR.is_dir(), f"Folder data tidak ditemukan: {DATA_DIR}"
print("Data:", DATA_DIR)
print("Mode cepat:", MODE_CEPAT)

def sigmoid(x):
    x = np.clip(x, -60, 60)
    return 1 / (1 + np.exp(-x))

def softmax(x):
    x = np.asarray(x)
    shifted = x - np.max(x, axis=-1, keepdims=True)
    exp = np.exp(shifted)
    return exp / exp.sum(axis=-1, keepdims=True)
```

```python
def sigmoid(x): return 1 / (1 + np.exp(-x))
def relu(x): return np.maximum(0, x)
def tanh(x): return np.tanh(x)
def softmax(x):
    shifted = x - np.max(x, axis=-1, keepdims=True)
    exp = np.exp(shifted)
    return exp / exp.sum(axis=-1, keepdims=True)

values = np.array([-3., -1., 0., 1., 3.])
print("sigmoid", np.round(sigmoid(values), 3))
print("tanh   ", np.round(tanh(values), 3))
print("relu   ", relu(values))
print("softmax", np.round(softmax(values), 3), "sum=", softmax(values).sum())
```

```python
# Cross-entropy dan gradien softmax
logits = np.array([[2.0, 0.5, -1.0]])
target = np.array([[1.0, 0.0, 0.0]])
prob = softmax(logits)
loss = -np.sum(target * np.log(prob + 1e-12))
grad_logits = prob - target
print("probability:", prob, "loss:", loss, "gradient:", grad_logits)
```
